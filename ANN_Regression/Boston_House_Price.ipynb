## 1. Import Libraries

```python
import numpy as np

from tensorflow.keras.datasets import boston_housing
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

## 2. Load Boston Housing Dataset

```python
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()
```

## 3. Check Dataset Shape

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

### 4. Data Preparation

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("Training mean:", X_train_scaled.mean().round())
print("Training std:", X_train_scaled.std().round())
```

```python
print("Test mean:", X_test_scaled.mean().round())
print("Test std:", X_test_scaled.std().round())
```

### 5. Model Building

```python
model = Sequential()

model.add(Dense(64, activation='relu', input_shape=(13,)))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1,))
```

```python
model.compile(optimizer='adam',loss='mse', metrics=['mae'])
```

### 6. Model Training || 7. Model Testing || 8. Model Evaluation

```python
history = model.fit(X_train_scaled, y_train, epochs=100, batch_size=16, validation_split=0.2, verbose=2)
```

```python
test_loss, test_mae = model.evaluate(X_test_scaled, y_test, verbose=0)

print("Test Loss (MSE):", test_loss)
print("Test MAE:", test_mae)
```

```python
model.save("boston_house_model.h5")
```

```python
import numpy as np
import joblib
from tensorflow.keras.models import load_model

# Load trained ANN model
model = load_model(
    "boston_house_model.h5",
    compile=False
)

# Load scaler used during training
scaler = joblib.load(
    "scaler.pkl"
)

print("Model and scaler loaded successfully.")
```

```python
def predict_house_price(features):

    # Convert the 13 input features into a 2D array
    data = np.array([features], dtype=np.float32)

    # Apply the same scaler used during training
    data_scaled = scaler.transform(data)

    # Generate regression prediction
    prediction = model.predict(
        data_scaled,
        verbose=0
    )

    # Return the single predicted value
    return float(prediction[0][0])
```

```python
features = [
    0.02,   # CRIM
    18,     # ZN
    2.5,    # INDUS
    0,      # CHAS
    0.45,   # NOX
    6.5,    # RM
    65,     # AGE
    4.5,    # DIS
    1,      # RAD
    300,    # TAX
    15,     # PTRATIO
    390,    # B
    5       # LSTAT
]

price = predict_house_price(features)

print("Predicted House Price:", price)
```
